Задание 1

Дана переменная, в которой хранится слово из латинских букв. Напишите код, который выводит на экран:
    среднюю букву, если число букв в слове нечетное;
    две средних буквы, если число букв четное.

Примеры работы программы:

    word = 'test'
Результат:
    es

    word = 'testing'
Результат:
    t

```python
word = input("Введите слово: ")
length = len(word)
result = ''
for i in range(length):
    # Для нечетной длины слова
    if length % 2 == 1 and i == length // 2:
        result = word[i]
    
    # Для четной длины слова
    if length % 2 == 0 and (i == length // 2 - 1 or i == length // 2):
        result += word[i]

print(result)
```

Задание 2

Напишите программу, которая последовательно запрашивает у пользователя числа (по одному за раз) и после первого нуля выводит сумму всех ранее введенных чисел.

Примеры работы программы:
1. Введите число:  1
   Введите число:  4
   Введите число:  6
   Введите число:  0

Результат: 11

2. Введите число:  0

Результат: 0

```python
sum = 0
n = int(input('Введите число'))

while n != 0:
    n = int(input('Введите число'))
    sum += n
    if n == 0:
        print(sum)
        break
```

Задание 3

Мы делаем MVP dating-сервиса, и у нас есть список парней и девушек.
Выдвигаем гипотезу: лучшие рекомендации мы получим, если просто отсортируем имена по алфавиту и познакомим людей с одинаковыми индексами после сортировки! Но мы не будем никого знакомить, если кто-то может остаться без пары:

Примеры работы программы:
1. boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
   girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

Результат:
Идеальные пары:  
    Alex и Emma  
    Arthur и Kate  
    John и Kira  
    Peter и Liza  
    Richard и Trisha 
    
2. boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
   girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

Результат:
    Внимание, кто-то может остаться без пары!

```python
def find_pair(boys, girls):
    # Если массивы имён не равны по длине
    if len(boys) != len(girls):
        print("Кто-то может остаться без пары, знакомств не будет!")
        return []
    
    # Если массивы не попали под условие выше 
    
    sorted_boys = sorted(boys)
    sorted_girls = sorted(girls)
    pairs_length = len(boys)
    pairs = []

    for i in range(pairs_length):
        pairs.append((sorted_boys[i], sorted_girls[i]))
    
    return pairs

boys = [name.strip() for name in input("Введите имена парней через запятую: ").split(',') if name.strip()]
girls = [name.strip() for name in input("Введите имена девушек через запятую: ").split(',') if name.strip()]

pairs = find_pair(boys, girls)

if pairs:
    print("Идеальные пары:")
    for boy, girl in pairs:
        print(f"{boy} и {girl}")
```

Задание 4

У нас есть список, содержащий информацию о среднедневной температуре в Фаренгейтах за произвольный период по странам (структура данных в примере). 
Необходимо написать код, который рассчитает среднюю температуру за период в Цельсиях(!) для каждой страны.

Пример работы программы:

countries_temperature = [
    ['Таиланд', [75.2, 77, 78.8, 73.4, 68, 75.2, 77]],
    ['Германия', [57.2, 55.4, 59, 59, 53.6]],
    ['Россия', [35.6, 37.4, 39.2, 41, 42.8, 39.2, 35.6]],
    ['Польша', [50, 50, 53.6, 57.2, 55.4, 55.4]]
]
Результат:

Средняя температура в странах:
Таиланд  -  23.9 С
Германия  -  13.8 С
Россия  -  3.7 С
Польша  -  12.0 С

```python
# Функция для конвертации из Цельсиуса в Фаренгейт
def fahrenheit_to_celsius(f_temp):
    return (f_temp - 32) * 5 / 9

# Функция для расчёта средней температуры
def calculate_average_celsius(temperatures_f):
    temperatures_c = [fahrenheit_to_celsius(temp) for temp in temperatures_f]
    total = 0
    for temp_c in temperatures_c:
        total += temp_c
    
    return total / len(temperatures_c)

# Функция для определение средняя температуры в странах
def process_countries_temperature(data):
    print("Средняя температура в странах:")
    for country, temps_f in data:
        avg_temp_c = calculate_average_celsius(temps_f)
        print(f"{country}  -  {avg_temp_c:.1f} С")

if __name__ == "__main__":
    countries_temperature = [
        ['Таиланд', [75.2, 77, 78.8, 73.4, 68, 75.2, 77]],
        ['Германия', [57.2, 55.4, 59, 59, 53.6]],
        ['Россия', [35.6, 37.4, 39.2, 41, 42.8, 39.2, 35.6]],
        ['Польша', [50, 50, 53.6, 57.2, 55.4, 55.4]]
    ]
    
    process_countries_temperature(countries_temperature)
```
